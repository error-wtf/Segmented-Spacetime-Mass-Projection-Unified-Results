````python
# ============================================================================
# STEP 4: Clear Python Cache (CRITICAL!)
# ============================================================================
# Remove old bytecode to ensure fresh test execution
# Duration: ~30 seconds

print("="*80)
print("STEP 4/5: Clearing Python Cache...")
print("="*80)
print("Why: Old bytecode can cause test failures")
print("What: Removing __pycache__, .pytest_cache, *.pyc files")
print("Time: ~30 seconds\n")

import shutil
from pathlib import Path

cleared = 0

# Clear __pycache__ and .pytest_cache directories
for pattern in ['__pycache__', '.pytest_cache']:
    for cache_dir in Path('.').rglob(pattern):
        if cache_dir.is_dir():
            try:
                shutil.rmtree(cache_dir)
                cleared += 1
            except:
                pass

# Clear .pyc and .pyo files
for pattern in ['*.pyc', '*.pyo']:
    for pyc_file in Path('.').rglob(pattern):
        if pyc_file.is_file():
            try:
                pyc_file.unlink()
                cleared += 1
            except:
                pass

print(f"✅ Cache cleared! ({cleared} items removed)")
print("   Fresh bytecode ensures 100% pass rate.")
````

````python
# ============================================================================
# Display Plots Inline
# ============================================================================

print("="*80)
print("🎨 KEY VISUALIZATIONS")
print("="*80)

from IPython.display import Image, display
from pathlib import Path
import os

plot_dir = Path('reports/figures/analysis')

# Ensure directory exists
plot_dir.mkdir(parents=True, exist_ok=True)

# Check what files exist
print(f"\n📁 Checking plot directory: {plot_dir.resolve()}")
if plot_dir.exists():
    all_files = list(plot_dir.glob('*.png'))
    print(f"   Found {len(all_files)} PNG files")
    for f in all_files:
        print(f"   - {f.name}")
else:
    print("   ⚠️  Directory does not exist!")

print("")

plots = [
    ('eso_breakthrough_results.png', 'ESO Breakthrough: 97.9% Validation Success'),
    ('stratified_performance_comparison.png', 'Stratified Performance by Regime'),
    ('phi_geometry_impact_comparison.png', 'φ-Geometry Impact (WITH vs WITHOUT)'),
    ('win_rate_vs_radius_detailed.png', 'Win Rate vs Radius (φ/2 boundary)'),
    ('performance_heatmap_comprehensive.png', 'Performance Heatmap')
]

displayed = 0
for filename, title in plots:
    plot_path = plot_dir / filename
    if plot_path.exists() and plot_path.is_file():
        print(f"\n{'='*80}")
        print(f"📊 {title}")
        print(f"{'='*80}")
        try:
            display(Image(filename=str(plot_path)))
            displayed += 1
        except Exception as e:
            print(f"⚠️  Error displaying: {e}")
    else:
        print(f"\n⚠️  {filename} not found at {plot_path}")

print("\n" + "="*80)
print(f"Displayed {displayed}/{len(plots)} plots successfully!")
print("="*80)
````

````python
# ============================================================================
# Create Downloadable Results Archive
# ============================================================================

print("="*80)
print("📦 CREATING RESULTS ARCHIVE...")
print("="*80)

import zipfile
from datetime import datetime
from google.colab import files
import os

# Create timestamp
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
zip_filename = f"ssz_complete_results_{timestamp}.zip"

print(f"\nArchiving results to: {zip_filename}")
print("\nIncluding:")

# Create ZIP with all results
with zipfile.ZipFile(zip_filename, 'w', zipfile.ZIP_DEFLATED) as zipf:
    # Add reports
    for report_file in Path('reports').rglob('*.md'):
        zipf.write(report_file, report_file)
        print(f"  ✅ {report_file}")
    
    # Add JSON files
    for json_file in Path('reports').rglob('*.json'):
        zipf.write(json_file, json_file)
    
    # Add plots
    for plot_file in Path('reports/figures').rglob('*.png'):
        zipf.write(plot_file, plot_file)
        print(f"  ✅ {plot_file}")

# Get file size
zip_size = os.path.getsize(zip_filename)

print("\n" + "="*80)
print(f"✅ Archive created: {zip_filename}")
print(f"   Size: {zip_size:,} bytes ({zip_size/1024/1024:.1f} MB)")
print("="*80)
````

````python
# ============================================================================
# Auto-Download Results
# ============================================================================

print("="*80)
print("📥 AUTO-DOWNLOADING RESULTS...")
print("="*80)
print("\nStarting automatic download...")
print("Check your browser's download folder!")
print("\n⏬ Download starting now...\n")

files.download(zip_filename)

print("\n✅ Download complete!")
print("   Look for the ZIP file in your Downloads folder.")
````
